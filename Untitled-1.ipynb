```python
class User:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Withdrawal Successful")
            print("Available Amount:", self.balance)
        else:
            print("Insufficient Balance")


name = input("Enter user name: ")
balance = float(input("Enter account balance: "))

user = User(name, balance)

amount = float(input("Enter withdrawal amount: "))
user.withdraw(amount)
```

```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


n = int(input("Enter a number: "))
print("Factorial =", factorial(n))
```

```python
def is_palindrome(s: str) -> bool:
    left = 0
    right = len(s) - 1

    while left < right:
        
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1

        
        if s[left].lower() != s[right].lower():
            return False

        left += 1
        right -= 1

    return True



print(is_palindrome("racecar"))  
print(is_palindrome("A man, a plan, a canal: Panama"))  
print(is_palindrome("hello"))  
```

```python
import openpyxl


def load_activity_log(file_path):
    workbook = openpyxl.load_workbook(file_path, data_only=True)
    worksheet = (
        workbook["Daily Log"] if "Daily Log" in workbook.sheetnames else workbook.active
    )

    daily_records_list = []
    for current_row in worksheet.iter_rows(min_row=7, values_only=True):
        entry_date = current_row[0]
        if entry_date is None:
            continue

        record_entry = {
            "date": entry_date,
            "sleep_minutes": current_row[1] if current_row[1] is not None else 0,
            "fitness_minutes": current_row[2] if current_row[2] is not None else 0,
            "study_minutes": current_row[3] if current_row[3] is not None else 0,
            "coding_minutes": current_row[4] if current_row[4] is not None else 0,
            "class_minutes": current_row[5] if current_row[5] is not None else 0,
            "classes_attended_count": (
                current_row[6] if current_row[6] is not None else 0
            ),
            "other_activities_minutes": (
                current_row[7] if current_row[7] is not None else 0
            ),
            "total_tracked_minutes": (
                current_row[8] if current_row[8] is not None else 0
            ),
            "free_unaccounted_minutes": (
                current_row[9] if current_row[9] is not None else 0
            ),
            "feeling_mood": current_row[10],
            "satisfaction_level": current_row[11],
            "energy_level": current_row[12],
        }
        daily_records_list.append(record_entry)

    return daily_records_list


def calculate_daily_averages(daily_records_list):
    tracked_days_count = 0
    total_sleep_minutes = 0
    total_fitness_minutes = 0
    total_study_minutes = 0
    total_coding_minutes = 0
    total_class_minutes = 0
    total_classes_attended = 0

    for record in daily_records_list:
        total_sleep_minutes = total_sleep_minutes + record["sleep_minutes"]
        total_fitness_minutes = total_fitness_minutes + record["fitness_minutes"]
        total_study_minutes = total_study_minutes + record["study_minutes"]
        total_coding_minutes = total_coding_minutes + record["coding_minutes"]
        total_class_minutes = total_class_minutes + record["class_minutes"]
        total_classes_attended = (
            total_classes_attended + record["classes_attended_count"]
        )
        tracked_days_count = tracked_days_count + 1

    if tracked_days_count == 0:
        return None

    return {
        "tracked_days": tracked_days_count,
        "avg_sleep_minutes": total_sleep_minutes / tracked_days_count,
        "avg_fitness_minutes": total_fitness_minutes / tracked_days_count,
        "avg_study_minutes": total_study_minutes / tracked_days_count,
        "avg_coding_minutes": total_coding_minutes / tracked_days_count,
        "avg_class_minutes": total_class_minutes / tracked_days_count,
        "avg_classes_attended": total_classes_attended / tracked_days_count,
    }


def analyze_friends_data(file_path="friends.xlsx"):
    workbook = openpyxl.load_workbook(file_path, data_only=True)
    worksheet = workbook.active

    total_age = 0
    total_height = 0
    total_weight = 0
    total_friends_count = 0

    for current_row in worksheet.iter_rows(min_row=2, values_only=True):
        if current_row[0] is None:
            continue

        friend_age = current_row[1] if current_row[1] is not None else 0
        friend_height = current_row[2] if current_row[2] is not None else 0
        friend_weight = current_row[3] if current_row[3] is not None else 0

        total_age = total_age + friend_age
        total_height = total_height + friend_height
        total_weight = total_weight + friend_weight
        total_friends_count = total_friends_count + 1

    if total_friends_count == 0:
        return None

    return {
        "total_friends": total_friends_count,
        "average_age": total_age / total_friends_count,
        "average_height": total_height / total_friends_count,
        "average_weight": total_weight / total_friends_count,
    }
```
